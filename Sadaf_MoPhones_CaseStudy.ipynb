```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
path = r"C:\Users\Administrator\Downloads\Credit Data\Credit Data"

files = [
    ("Credit Data - 01-01-2025.csv", "2025-01-01"),
    ("Credit Data - 30-03-2025.csv", "2025-03-30"),
    ("Credit Data - 30-06-2025.csv", "2025-06-30"),
    ("Credit Data - 30-09-2025.csv", "2025-09-30"),
    ("Credit Data - 30-12-2025.csv", "2025-12-30")
]

df_list = []

for file, date in files:
    temp = pd.read_csv(path + "\\" + file)
    temp['snapshot_date'] = pd.to_datetime(date)
    df_list.append(temp)

credit_df = pd.concat(df_list, ignore_index=True)

credit_df.head()
```

```python
customer = pd.read_excel(r"C:\Users\Administrator\Downloads\Sales and Customer Data\Sales and Customer Data.xlsx")

customer.head()
```

```python
nps = pd.read_excel(r"C:\Users\Administrator\Downloads\NPS Data.xlsx")

nps.head()
```

```python
credit_df.columns = credit_df.columns.str.strip().str.lower()
customer.columns = customer.columns.str.strip().str.lower()
nps.columns = nps.columns.str.strip().str.lower()
```

```python
customer.rename(columns={'loan id': 'loan_id'}, inplace=True)
nps.rename(columns={'loan id': 'loan_id'}, inplace=True)
```

```python
#Merge Data
df = credit_df.merge(customer, on='loan_id', how='left')
df = df.merge(nps, on='loan_id', how='left')

df.head()
```

```python
#Feature Engineering
# Default & arrears
df['is_default'] = df['days_past_due'] > 30
df['is_arrears'] = df['arrears'] > 0
```

```python
# NPS rename
df.rename(columns={
    'using a scale from 0 (not likely) to 10 (very likely), how likely are you to recommend mophones to friends or family?': 'nps_score'
}, inplace=True)
```

```python
# NPS category
def nps_category(x):
    if x >= 9:
        return 'Promoter'
    elif x >= 7:
        return 'Passive'
    else:
        return 'Detractor'

df['nps_category'] = df['nps_score'].apply(nps_category)
```

```python
# Age band
bins = [18,25,35,45,55,100]
labels = ['18-25','26-35','36-45','46-55','55+']
df['age_band'] = pd.cut(df['customer_age'], bins=bins, labels=labels)
```

```python
# Missing values fix
df['age_band'] = df['age_band'].astype(str).fillna('Unknown')
df['nps_score'] = df['nps_score'].fillna(df['nps_score'].median())
```

```python
#Portfolio Metrics (Q1)
portfolio = df.groupby('snapshot_date').agg({
    'is_default': 'mean',
    'is_arrears': 'mean',
    'balance': 'mean'
}).reset_index()

portfolio.rename(columns={
    'is_default': 'default_rate',
    'is_arrears': 'arrears_rate'
}, inplace=True)

portfolio
```

```python
#Default Rate Graph
plt.figure()
plt.plot(portfolio['snapshot_date'], portfolio['default_rate'], marker='o')
plt.title("Default Rate Over Time")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(r"C:\Users\Administrator\Downloads\default_rate.png")
plt.show()
```

```python
#Arrears Graph
plt.figure()
plt.plot(portfolio['snapshot_date'], portfolio['arrears_rate'], marker='o')
plt.title("Arrears Rate Over Time")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(r"C:\Users\Administrator\Downloads\arrears_rate.png")
plt.show()
```

```python
#Age Segment Analysis
age_risk = df.groupby('age_band')['is_default'].mean().sort_values(ascending=False)

age_risk.plot(kind='bar')
plt.title("Default Rate by Age Band")

plt.savefig(r"C:\Users\Administrator\Downloads\age_segment.png")
plt.show()

age_risk
```

```python
#NPS vs Default (Q2)
plt.figure()
sns.boxplot(x='is_default', y='nps_score', data=df)

plt.title("NPS Score vs Default Status")

plt.savefig(r"C:\Users\Administrator\Downloads\nps_boxplot.png")
plt.show()
```
